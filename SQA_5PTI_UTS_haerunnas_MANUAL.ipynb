```python
# PERHITUNGAN GAJI
print("=== PROGRAM PERHITUNGAN GAJI JOHN TRAVOLTA ===")

# Input data
jam_kerja = int(input("Masukkan jumlah jam kerja dalam 1 minggu : "))
rate_normal = 15000
jam_normal = 40

# Proses perhitungan gaji
if jam_kerja <= jam_normal:
    gaji = jam_kerja * rate_normal
    gaji_lembur = 0
else:
    jam_lembur = jam_kerja - jam_normal
    gaji_normal = jam_normal * rate_normal
    rate_lembur = rate_normal * 1.5
    gaji_lembur = jam_lembur * rate_lembur
    gaji = gaji_normal + gaji_lembur


# OUTPUT GAJI
print("\n--- HASIL PERHITUNGAN GAJI ---")
print("Jam kerja       :", jam_kerja, "jam")
print("Gaji normal     : Rp", int(gaji - gaji_lembur))
print("Gaji lembur     : Rp", int(gaji_lembur))
print("Total gaji      : Rp", int(gaji))



# PERHITUNGAN TABUNGAN
print("\n=== PROGRAM PERHITUNGAN TABUNGAN ===")

# Input pengeluaran
pengeluaran = int(input("Masukkan pengeluaran selama 1 minggu : "))

# Proses perbandingan pemasukan dan pengeluaran
if gaji > pengeluaran:
    tabungan = gaji - pengeluaran
    status = "Bisa menabung"
elif gaji == pengeluaran:
    tabungan = 0
    status = "Tidak bisa menabung"
else:
    tabungan = 0
    status = "Cari tambahan"



# OUTPUT KONDISI KEUANGAN
print("\n--- HASIL KEUANGAN ---")
print("Pemasukan       : Rp", int(gaji))
print("Pengeluaran     : Rp", pengeluaran)
print("Status          :", status)

if tabungan > 0:
    print("Jumlah tabungan : Rp", int(tabungan))
```
